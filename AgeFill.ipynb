```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import os

V_T2 = pd.read_csv("Data//Standardised_IDs//vandenBerg_table2.csv")        
K_21 = pd.read_csv("Data//Standardised_IDs//Krause21.csv")
```

```python
AGES = pd.merge(K_21[["ID","Age"]],V_T2[["ID","Age"]],on="ID",how="outer",suffixes=["K","V"])
AGES = AGES.fillna("NA")
#AGES
AGES["Merged"]=AGES["AgeV"]
AGES["Merged"].loc[AGES["AgeV"]=="NA"]=AGES["AgeK"]
#AGES["Merged"]

#print(AGES["Merged"].unique())
#print(V_T2["Age"].fillna(K_21["Age"]).unique())

MergedAges=pd.DataFrame()
MergedAges["ID"]=AGES["ID"]
MergedAges["Age"]=AGES["Merged"]



#========= Read old Merged DF
DirName = "MetalicityMerge"
FileName = "MergedMetalicities"
path = "Data//"+str(DirName)+"//"+str(FileName)+".csv"

DF=pd.read_csv(path)

#========= Add ages and write to file
MergedMetalicities=pd.merge(DF,MergedAges,on="ID",how="outer")

if os.path.exists(path):
    yn = input("The file '%s' already exists. Overwrite [y/n]?"%FileName)
    if yn == "y":
        MergedMetalicities.to_csv(path)
else:   MergedMetalicities.to_csv(path)
```
