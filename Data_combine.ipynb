```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
harris_p1 = pd.read_csv('HarrisPartI.csv')
harris_p3 = pd.read_csv('HarrisPartIII.csv')
vdB = pd.read_csv('vandenBerg.csv')
krause = pd.read_csv('Krause21.csv')


harris_p1['ID'] = harris_p1['ID'].str.replace(" ", "")
harris_p3['ID'] = harris_p3['ID'].str.replace(" ", "")
vdB['ID'] = vdB['ID'].str.replace(" ", "")
krause['ID'] = krause['ID'].str.replace(" ", "")

combined_data = pd.merge(harris_p1, harris_p3, on='ID', how='outer')
combined_data = pd.merge(combined_data, vdB, on='ID', how='outer')
combined_data = pd.merge(combined_data, krause, on='ID', how='outer')
combined_data.to_csv('Combined_Data.csv', index=False)
```
